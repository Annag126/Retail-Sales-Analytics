```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Superstore.csv", encoding="cp1252")
```

```python
print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Missing values:")
display(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Data types:")
display(df.dtypes)
```

```python
print("Number of rows and columns:", df.shape)
```

```python
print("Exact duplicate rows:", df.duplicated().sum())
```

```python
print("Unique Order IDs:", df["Order ID"].nunique())
```

```python
print("Unique Products:", df["Product ID"].nunique())
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"], dayfirst=True, errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], dayfirst=True, errors="coerce"
)

print(df[["Order Date", "Ship Date"]].dtypes)

print("Invalid Order Dates:", df["Order Date"].isna().sum())
print("Invalid Ship Dates:", df["Ship Date"].isna().sum())
```

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.strftime("%B")
df["Order Year-Month"] = df["Order Date"].dt.to_period("M").astype(str)

df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

display(df[[
    "Order Date",
    "Order Year",
    "Month Name",
    "Order Year-Month",
    "Sales",
    "Profit",
    "Shipping Days"
]].head())
```

```python
df.columns = df.columns.str.strip().str.replace(" ", "_")

print(df.columns.tolist())
```

```python
df.to_csv("Superstore_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
print("Rows and columns:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

display(df.head())
```

```python
monthly_sales = (
    df.groupby("Order_Year-Month")["Sales"]
    .sum()
    .reset_index()
    .sort_values("Order_Year-Month")
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales["Order_Year-Month"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend", fontsize=14)
plt.xlabel("Year-Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig("monthly_sales_trend.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

category_sales.plot(kind="bar")

plt.title("Total Sales by Category", fontsize=14)
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("sales_by_category.png", dpi=300, bbox_inches="tight")
plt.show()

display(category_sales)
```

```python
category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
category_profit.plot(kind="bar")

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.axhline(y=0, linewidth=0.8)
plt.tight_layout()

plt.savefig("profit_by_category.png", dpi=300, bbox_inches="tight")
plt.show()

display(category_profit)
```

```python
top_products = (
    df.groupby("Product_Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .sort_values()
)

plt.figure(figsize=(11, 7))
top_products.plot(kind="barh")

plt.title("Top 10 Products by Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.tight_layout()

plt.savefig("top_10_products_by_sales.png", dpi=300, bbox_inches="tight")
plt.show()

display(top_products.sort_values(ascending=False))
```

```python
region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
region_sales.plot(kind="bar")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("sales_by_region.png", dpi=300, bbox_inches="tight")
plt.show()

display(region_sales)
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(7, 7))
plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Share by Category")
plt.tight_layout()
plt.savefig("sales_share_pie.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.hist(df["Sales"], bins=30, edgecolor="black")

plt.title("Distribution of Sales")
plt.xlabel("Sales Amount")
plt.ylabel("Number of Transactions")
plt.tight_layout()

plt.savefig("sales_histogram.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
numeric_data = df[
    ["Sales", "Quantity", "Discount", "Profit", "Shipping_Days"]
]

correlation_matrix = numeric_data.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.tight_layout()

plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    df["Sales"],
    df["Profit"],
    alpha=0.4
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.axhline(y=0, linestyle="--")
plt.tight_layout()

plt.savefig("sales_vs_profit.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Business Metrics and EDA Summary**

```python
# Overall business metrics
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()
total_orders = df["Order_ID"].nunique()
average_order_sales = df.groupby("Order_ID")["Sales"].sum().mean()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Total Orders: {total_orders:,}")
print(f"Average Sales per Order: ${average_order_sales:,.2f}")
```

```python
print("Sales by Category:")
display(
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .round(2)
)

print("Profit by Category:")
display(
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .round(2)
)

print("Sales by Region:")
display(
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .round(2)
)

print("Profit by Region:")
display(
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .round(2)
)
```

```python
product_profit = (
    df.groupby("Product_Name")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .sort_values("Total_Profit")
)

print("10 Products with the Lowest Total Profit:")
display(product_profit.head(10).round(2))
```

**Business Insights and Recommendations**

**Business Insights**

1. Overall Performance: Total sales were USD 2,297,200.86, with a profit of USD 286,397.02 across 5,009 orders.

2. Category Sales: Technology recorded the highest sales at USD 836,154.03, followed by Furniture at USD 741,999.80 and Office Supplies at USD 719,047.03.

3. Category Profitability: Technology generated the highest profit of USD 145,454.95. Furniture generated only USD 18,451.27 in profit despite having the second-highest sales.

4. Regional Sales: The West region recorded the highest sales at USD 725,457.82, followed by the East at USD 678,781.24. The South had the lowest sales at USD 391,721.90.

5. Regional Profit: The West generated the highest profit of USD 108,418.45, while the Central region recorded the lowest at USD 39,706.36.

6. Product Profitability: The Cubify CubeX 3D Printer Double Head Print generated sales of USD 11,099.96 but recorded a loss of USD 8,879.97.

**Recommendations**

1. Review Furniture pricing, costs, and discounts to understand its comparatively low profit.

2. Investigate loss-making products and review their pricing, discounts, and costs.

3. Explore targeted promotions and product strategies to improve sales in the South region.

4. Analyze high-performing Technology products to identify opportunities for inventory planning and marketing.
